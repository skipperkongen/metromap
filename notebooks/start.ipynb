```python
import networkx as nx
import plotly.graph_objects as go
```

```python
def get_cph_metro():
    G = nx.MultiDiGraph()
    # Stations
    G.add_node(1, station='Vanløse', lon=12.491746694748002, lat=55.68709189133337)
    G.add_node(2, station='Flintholm St', lon=12.499102593325714, lat=55.685727760901464) 
    G.add_node(3, station='Lindevang St', lon=12.513057431301462, lat=55.683740529735914)
    # Lines
    G.add_edge(1, 2, line='M1')
    G.add_edge(2, 3, line='M1')
    G.add_edge(1, 2, line='M2')
    G.add_edge(2, 3, line='M2')
    return G
```

```python
def draw_geograpical_map(G: nx.DiGraph):
    fig = go.Figure()
    # Draw lines
    for u, v, data in G.edges(data=True):
        u_data = G.nodes[u]
        v_data = G.nodes[v]
        fig.add_trace(
            go.Scattergeo(
                lat = (u_data['lat'], v_data['lat']),
                lon = (u_data['lon'], v_data['lon']),
                mode = 'lines',
                line = dict(width = 1,color = 'red'),
            )
        )

    # Draw stations
    lat, lon, station = zip(*[(data['lat'], data['lon'], data['station']) for _, data in G.nodes(data=True)])
    fig.add_trace(go.Scattergeo(
        lat = lat,
        lon = lon,
        hoverinfo = 'text',
        text = station,
        mode = 'markers+text',
        textposition='bottom right',
        marker = dict(
            size = 10,
            color = 'rgb(255, 0, 0)',
            line = dict(
                width = 5,
                color = 'rgba(68, 68, 68, 0)'
            )
        )))
    fig.update_geos(
        fitbounds='locations',
        projection_type="equirectangular",
        landcolor="LightGreen",
        oceancolor="LightBlue",
        showocean=True,
        lakecolor="LightBlue"
    )    
    fig.update_layout(
        title_text = 'Feb. 2011 American Airline flight paths<br>(Hover for airport names)',
        showlegend = False,
        width=1200,
        height=1200,
        geo = dict(
            scope = 'europe',
            projection_type = 'azimuthal equal area',
            showland = True,
            landcolor = 'rgb(243, 243, 243)',
            countrycolor = 'rgb(204, 204, 204)',
        ),
    )
    
    fig.show()
```

```python
G = get_cph_metro()
draw_geograpical_map(G)
```

```python
def snap_to_grid(x, y, grid_size):
    snapped_x = round(x / grid_size) * grid_size
    snapped_y = round(y / grid_size) * grid_size
    return snapped_x, snapped_y
```
